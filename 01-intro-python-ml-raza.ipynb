##### Pivot wider data

```python
df = data.pivot(index = ['DATE', 'Block', 'Plot', 'Rotation', 'Quadrat', 'Outcome'], 
                columns = 'Wavelength',
                values = 'Reflectance').reset_index().rename_axis(None, axis=1)

# Adding string ('W') before wavelength band
new_names = {460: 'W460', 510: 'W510', 560: 'W560', 610: 'W610', 660: 'W660', 710: 'W710', 760: 'W760', 810: 'W810'}
df.rename(columns = new_names, inplace = True)

df
```

##### NDVI calculation

```python
df['NDVI1'] = (df['W760'] - df['W660']) / (df['W760'] + df['W660'])
df['NDVI2'] = (df['W810'] - df['W710']) / (df['W810'] + df['W710'])
df
```

## Data cleaning

### Recoding and changing data type 

```python
# function to recode crop rotation to a numeric number
def tran_Rotation(x):
  if x == 'S2':
    return 2
  if x == 'S3':
    return 3
  if x == 'S4':
    return 4
  
df['ROTATION'] = df['Rotation'].apply(tran_Rotation)
df['ROTATION'] = df['ROTATION'].astype('category') # making rotaton data as of category type

# top 10 rows of data
df.head(10)
```

### Filter data for a specific date

```python
# lets filter data for September 02 2016
df_n = df[df['DATE'] == '02/09/2016']
df_n.head()
```

## Define Explanatory and Response variables

```python
# create separate data frames for Explanatory and Response variables:
expl_vars = ['W460', 'W510', 'W560', 'W610', 'W660', 'W710', 'W760', 'W810', 'NDVI1', 'NDVI2', 'ROTATION']
resp_var = "Outcome"
print("Predicting", resp_var , "from", expl_vars)

# selecting explainatory and response variables from data frame
X = df_n[expl_vars] # Explanatory variables
y = df_n[resp_var]  # Response variable
```

## Data splitting

* Training data = 70%
* Test data = 30%

```python
from sklearn.model_selection import train_test_split 

# Split dataset into training set and test set
SPLIT_RND_SEED = 123
X_train, X_test, y_train, y_test = train_test_split(X, y,
                                                    test_size=0.3, 
                                                    random_state=SPLIT_RND_SEED)  # 70% training and 30% test

print("Using", len(X_train), "quadrants for training,", len(y_test), "quadrants for testing")
```

```python
# check the shape of X_train and X_test
X_train.shape, X_test.shape
```

## Random Forest classifier

### Model building

```python
# Import Random Forest Model
from sklearn.ensemble import RandomForestClassifier

# Create a Gaussian Classifier with 500 trees 
rf = RandomForestClassifier(n_estimators = 500, 
                                   oob_score=True, 
                                   random_state=12345, # random number to be used, needed to reproduce the same result
                                   verbose=False)
```

### Model fitting

```python
# Train the model using the training sets
rf_model = rf.fit(X_train, y_train)

# printing rhe classifier object shows its parameters
print(rf_model)
```

### Out of bag accuracy

```python
# Out of bag score
print(f'Out-of-bag score estimate: {rf.oob_score_:.3}')
```

### Model prediction

```python
# Prediction on test data
y_pred = rf.predict(X_test)
```

### Accuracy assessment

#### Confusion matrix

confusion matrix.jpg

#### Accuracy vs. Precision

image.png

```python
# Confusion (error) Matrix of Prediction
from sklearn.metrics import confusion_matrix

cm = pd.DataFrame(confusion_matrix(y_pred, y_test))
  
# use seaborn to plot matrix as heatmap    
plt.rc('font', size=16)
p = sns.heatmap(cm,
                annot=True,
                cbar=False,
                cmap="Oranges")
plt.xlabel('Ground Truth SDS')
plt.ylabel('Predicted SDS')
```

#### Model accuracy

```python
# Accuracy assessment
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)
print('Model accuracy is {:.3f}'.format(accuracy))
```

#### Classification report

```python
from sklearn.metrics import classification_report 

# Classification report
stats = classification_report(y_test, y_pred,
                              labels=None,
                              target_names=["Healty", "SDS"],
                              sample_weight=None,
                              digits=2,
                              output_dict=False)
  
print("Classification report:\n")
print(stats) 
```

#### ROC curve

A receiver operating characteristic curve, or ROC curve, is a graphical plot that illustrates the diagnostic ability of a binary classifier system as its discrimination threshold is varied. 

Roc_curve1.png

```python
from sklearn.metrics import RocCurveDisplay

# display the predictions
plt.rcParams['figure.figsize'] = [10, 10]
RocCurveDisplay.from_estimator(rf, X_test, y_test)
```

#### Kappa statistics

* The kappa statistic takes into account chance agreement

$$\frac{Observed\ agreement - Expected\ agreement}{1 - Expected\ agreement}$$

* When two measurements agree only at the chance level, the value of kappa is zero

* When the two measurements agree perfectly, the value of kappa is 1

* So, higher the better

```python
from sklearn.metrics import cohen_kappa_score

#def kappa_statistics(X_test, y_test): 
y_pred = rf.predict(X_test)    
cohen_score = cohen_kappa_score(y_test, y_pred)    
print("Kappa score:", cohen_score)
```

### Variable importance

```python
# view the feature scores
feature_scores = pd.Series(rf.feature_importances_, index=X_train.columns).sort_values(ascending=False)
feature_scores

# display the predictions
plt.rcParams['figure.figsize'] = [10, 10]

palette = sns.color_palette("Set2")

# Creating a seaborn bar plot
sns.barplot(x=feature_scores, y=feature_scores.index, palette=palette)
plt.title("Visualizing Important Features")
plt.xlabel('Feature Importance Score')
plt.ylabel('Features')

plt.show()
```
